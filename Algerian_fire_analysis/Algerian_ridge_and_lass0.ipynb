```python
classlabel =['fire','not fire']
plt.figure(figsize=(12,6))
plt.pie(percentage,labels=classlabel,autopct='%1.1f%%')
plt.title("Pie chart of Classes ")
plt.show()
```

```python
df_copy['Classes  ']=np.where(df_copy['Classes  '].str.contains('not fire'),'not fire','fire')
```

```python
##monthly anlaysis
df_temp=df_copy[df_copy['Region']==0]
plt.figure(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes  ',data=df_temp)
plt.xlabel('months')
plt.ylabel('number of fires')
plt.title('Fire Analysis of Bejaia Region ',fontweight='bold')
```

```python
df_temp=df_copy[df_copy['Region']==1]
plt.figure(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes  ',data=df_temp)
plt.xlabel('months')
plt.ylabel('number of fires')
plt.title('Fire Analysis of sidi Bel Region ',fontweight='bold')
```
